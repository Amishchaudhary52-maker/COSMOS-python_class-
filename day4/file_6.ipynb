```python
import pandas as pd
data={
    "name":["Amish","sudan","Anan","alok"],
    "score":[90,89,89,89]
}
df=pd.DataFrame(data)
print(df)
top_score=df["score"].max()
print(f"Top score is: {top_score}")
df["passed"]=df["score"] >=50
num_passed = df["passed"].sum()
print(f"passed: {num_passed}")
```
